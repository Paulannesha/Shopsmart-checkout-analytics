```python
import pandas as pd
import numpy as np
```

```python
np.random.seed(42)
N_USERS=12000
```

```python
START_DATE=pd.Timestamp("2026-01-01")
END_DATE=pd.Timestamp("2026-06-30")
```

```python
date_range_days=(END_DATE-START_DATE).days
signup_offsets= np.random.triangular(0,date_range_days, date_range_days, N_USERS).astype(int)
signup_dates= START_DATE + pd.to_timedelta(signup_offsets, unit="D")
```

```python
signup_dates[:5]
```

```python
variant=np.random.choice(["control","treatment"],size=N_USERS, p=[0.5,0.5])
```

```python
pd.Series(variant).value_counts()
```

```python
visited_site= np.ones(N_USERS, dtype=int)

add_to_cart_rate=np.where(variant=="treatment",0.42,0.40)
added_to_cart=np.random.binomial(1,add_to_cart_rate)
```

```python
pd.Series(added_to_cart).mean()
```

```python
Start_checkout_rate=np.where(variant=="treatment",0.65,0.55)
started_checkout=added_to_cart*np.random.binomial(1,Start_checkout_rate)

complete_purchase_rate=np.where(variant=="treatment",0.58,0.46)
completed_purchase=started_checkout*np.random.binomial(1,complete_purchase_rate)
```

```python
pd.DataFrame({"variant":variant,"completed_purchase": completed_purchase}).groupby("variant")["completed_purchase"].mean()
```

```python
revenue=np.where(completed_purchase==1,np.round(np.random.lognormal(mean=4.0,sigma=0.5,size=N_USERS),2),0.0)
```

```python
pd.DataFrame({"variant":variant,"revenue":revenue}).groupby("variant")["revenue"].mean()
```

```python
month_1_active=np.random.binomial(1,np.where(variant=="treatment",0.38,0.32))
month_2_active=month_1_active*np.random.binomial(1,np.where(variant=="treatment",0.55,0.48))
month_3_active=month_2_active*np.random.binomial(1,np.where(variant=="treatment",0.60,0.52))
```

```python
pd.DataFrame({"variant":variant,"m1":month_1_active,"m2":month_2_active,"m3":month_3_active}).groupby("variant").mean()
```

```python
df=pd.DataFrame({"user_id":[f"U{100000+i}"for i in range(N_USERS)],
                 "signup_date":signup_dates,
                 "cohort_month":signup_dates.to_period("M").astype(str),
                                                       "variant":variant,"visited_site":visited_site,"added_to_cart":added_to_cart,
                 "started_checkout":started_checkout,"completed_purchase":completed_purchase,"revenue":revenue,"month_1_active":month_1_active,
                 "month_2_active":month_1_active,"month_3_active":month_1_active})
                                                       
```

```python
df.to_csv("../data/shopsmart_experiment.csv",index=False)
```
