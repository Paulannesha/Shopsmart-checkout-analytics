```python
from scipy import stats
```

```python
import pandas as pd
df=pd.read_csv("../data/shopsmart_experiment.csv")
df.head()
```

```python

control=df[df["variant"]=="control"]
treatment=df[df["variant"]=="treatment"]
```

```python
count=[treatment["completed_purchase"].sum(),control["completed_purchase"].sum()]
nobs=[len(treatment),len(control)]

from statsmodels.stats.proportion import proportions_ztest
z_stat,p_value=proportions_ztest(count,nobs)

print("Z-statistics:",z_stat)
print("P-value:",p_value)
```

Conducted a two proportion z-test on 12000 simulated users, finding the checkout redesign improved conversion by 52% (10.0% -> 15.2%), a statistically significant result (p<0.001)
